```python
#cada sabor es 1 si la fila lo contiene, 0 si no
naranja = df["Sabor"].str.contains("Naranja").astype(int)
coco = df["Sabor"].str.contains("Coco").astype(int)
khalua = df["Sabor"].str.contains("Khalua").astype(int)
fresa_pina = df["Sabor"].str.contains("Fresa").astype(int)
sandia = df["Sabor"].str.contains("Sandia").astype(int)

# Tamaño: Chica = 0, Grande = 1    Hielo: No = 0, Si = 1
tamano = (df["Tamaño"] == "Grande").astype(int)
hielo = (df["Hielo"] == "Si").astype(int)
```

```python
tabla = pd.DataFrame({
    "Naranja": naranja,
    "Coco": coco,
    "Khalua": khalua,
    "Fresa-Piña": fresa_pina,
    "Sandia": sandia,
    "Tamaño": tamano,
    "Hielo": hielo,
    "Calificacion": df["Calificacion"],
})

tabla
```

```python
tabla.dtypes
```

```python
tabla.describe()
```

```python
X = tabla.drop(columns="Calificacion")
X.insert(0, "Unos", 1)

y = tabla["Calificacion"]

X.head()
```

```python
modelo = sm.OLS(y, X).fit()
print(modelo.summary())
```

# Usando OneHotEncoder y OrdinalEncoder

```python
from sklearn.preprocessing import OrdinalEncoder, OneHotEncoder

# OneHotEncoder para el sabor: una columna 0/1 por cada sabor
ohe = OneHotEncoder(categories=[["Naranja", "Coco", "Khalua", "Fresa-Piña", "Sandia"]], dtype=int)
sabores = ohe.fit_transform(df[["Sabor"]]).toarray()

ohe.get_feature_names_out()
```

```python
# OrdinalEncoder para Tamaño y Hielo: Chica/No = 0, Grande/Si = 1
oe = OrdinalEncoder(categories=[["Chica", "Grande"], ["No", "Si"]], dtype=int)
ordinales = oe.fit_transform(df[["Tamaño", "Hielo"]])

oe.categories_
```

```python
tabla_sk = pd.DataFrame(sabores, columns=ohe.categories_[0], index=df.index)
tabla_sk[["Tamaño", "Hielo"]] = ordinales
tabla_sk["Calificacion"] = df["Calificacion"]

tabla_sk
```

```python
# X: columna de 1s, los 5 sabores, Tamaño y Hielo (sin Calificacion)
X_sk = tabla_sk.drop(columns="Calificacion")
X_sk.insert(0, "Unos", 1)

y_sk = tabla_sk["Calificacion"]

X_sk.head()
```

```python
modelo_sk = sm.OLS(y_sk, X_sk).fit()
print(modelo_sk.summary())
```

Calificacion = 5.4583 + 1.4917*Naranja + 0.6167*Coco + 2.3667*Khalua + 1.1167*Fresa-Piña - 0.1333*Sandia + 0.5000*Tamaño + 1.6000*Hielo
